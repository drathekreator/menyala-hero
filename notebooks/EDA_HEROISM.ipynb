```python
fitur_numerik = [
    "Usia",
    "IF_Pre",
    "OMD_Pre",
    "Potensi_Pre",
    "Eksplorasi_Pre",
    "Refleksi_Pre",
    "Perencanaan_Pre"
]

fitur_kategorikal = [
    "Jenis_Kelamin",
    "Terlibat_Memulung"
]

print("Fitur numerik:", len(fitur_numerik))
print("Fitur kategorikal:", len(fitur_kategorikal))
print("Total fitur:", len(fitur_numerik) + len(fitur_kategorikal))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

print("Library preprocessing berhasil diimport!")
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerik",
            StandardScaler(),
            fitur_numerik
        ),
        (
            "kategorikal",
            OneHotEncoder(handle_unknown="ignore"),
            fitur_kategorikal
        )
    ]
)

print("Preprocessor berhasil dibuat!")
```

```python
from iterstrat.ml_stratifiers import MultilabelStratifiedKFold

print("MultilabelStratifiedKFold berhasil diimport!")
```

```python
cv = MultilabelStratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Multilabel Stratified CV berhasil dibuat!")
```

```python
for fold, (train_idx, val_idx) in enumerate(
    cv.split(X, y_encoded),
    start=1
):
    print(
        f"Fold {fold}: "
        f"Training = {len(train_idx)}, "
        f"Validation = {len(val_idx)}"
    )
```

```python
for fold, (train_idx, val_idx) in enumerate(
    cv.split(X, y_encoded),
    start=1
):
    print(f"\n=== FOLD {fold} ===")

    y_val = y_encoded.iloc[val_idx]

    print("Jumlah data validation:", len(y_val))
    print("Jumlah label Perlu (1):")
    print(y_val.sum())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.multiclass import OneVsRestClassifier

print("Semua algoritma berhasil diimport!")
```

```python
model_lr = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", OneVsRestClassifier(
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    ))
])

model_dt = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", OneVsRestClassifier(
        DecisionTreeClassifier(
            random_state=42
        )
    ))
])

model_rf = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", OneVsRestClassifier(
        RandomForestClassifier(
            n_estimators=100,
            random_state=42
        )
    ))
])

print("3 pipeline model berhasil dibuat!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    hamming_loss
)

print("Metrik evaluasi berhasil diimport!")
```

```python
hasil_lr = []

for fold, (train_idx, val_idx) in enumerate(
    cv.split(X, y_encoded),
    start=1
):
    X_train = X.iloc[train_idx]
    X_val = X.iloc[val_idx]

    y_train = y_encoded.iloc[train_idx]
    y_val = y_encoded.iloc[val_idx]

    # Training
    model_lr.fit(X_train, y_train)

    # Prediksi
    y_pred = model_lr.predict(X_val)

    # Evaluasi
    accuracy = accuracy_score(y_val, y_pred)

    precision = precision_score(
        y_val,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_val,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        y_pred,
        average="macro",
        zero_division=0
    )

    hamming = hamming_loss(y_val, y_pred)

    hasil_lr.append([
        fold,
        accuracy,
        precision,
        recall,
        f1,
        hamming
    ])
```

```python
hasil_lr_df = pd.DataFrame(
    hasil_lr,
    columns=[
        "Fold",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "Hamming_Loss"
    ]
)

hasil_lr_df
```

```python
import numpy as np

y_true_all = []
y_pred_all = []

for train_idx, val_idx in cv.split(X, y_encoded):
    
    X_train = X.iloc[train_idx]
    X_val = X.iloc[val_idx]

    y_train = y_encoded.iloc[train_idx]
    y_val = y_encoded.iloc[val_idx]

    model_rf.fit(X_train, y_train)

    y_pred = model_rf.predict(X_val)

    y_true_all.append(y_val.values)
    y_pred_all.append(y_pred)

y_true_all = np.vstack(y_true_all)
y_pred_all = np.vstack(y_pred_all)

print("Ukuran aktual :", y_true_all.shape)
print("Ukuran prediksi:", y_pred_all.shape)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

nama_aspek = [
    "Potensi",
    "Eksplorasi",
    "Refleksi",
    "Perencanaan"
]

hasil_per_label = []

for i, aspek in enumerate(nama_aspek):
    
    precision = precision_score(
        y_true_all[:, i],
        y_pred_all[:, i],
        zero_division=0
    )
    
    recall = recall_score(
        y_true_all[:, i],
        y_pred_all[:, i],
        zero_division=0
    )
    
    f1 = f1_score(
        y_true_all[:, i],
        y_pred_all[:, i],
        zero_division=0
    )
    
    hasil_per_label.append([
        aspek,
        precision,
        recall,
        f1
    ])

hasil_per_label_df = pd.DataFrame(
    hasil_per_label,
    columns=[
        "Aspek",
        "Precision",
        "Recall",
        "F1"
    ]
)

hasil_per_label_df.round(3)
```

```python
parameter_rf = [
    {"n_estimators": 100, "max_depth": None, "min_samples_leaf": 1},
    {"n_estimators": 200, "max_depth": None, "min_samples_leaf": 1},
    {"n_estimators": 200, "max_depth": 5, "min_samples_leaf": 1},
    {"n_estimators": 200, "max_depth": 5, "min_samples_leaf": 2},
    {"n_estimators": 300, "max_depth": 5, "min_samples_leaf": 2},
    {"n_estimators": 300, "max_depth": 7, "min_samples_leaf": 2}
]
```

```python
hasil_tuning = []

for param in parameter_rf:

    f1_folds = []
    hamming_folds = []

    for train_idx, val_idx in cv.split(X, y_encoded):

        X_train = X.iloc[train_idx]
        X_val = X.iloc[val_idx]

        y_train = y_encoded.iloc[train_idx]
        y_val = y_encoded.iloc[val_idx]

        model_tuning = Pipeline([
            ("preprocessor", preprocessor),

            ("classifier", OneVsRestClassifier(
                RandomForestClassifier(
                    n_estimators=param["n_estimators"],
                    max_depth=param["max_depth"],
                    min_samples_leaf=param["min_samples_leaf"],
                    random_state=42
                )
            ))
        ])

        model_tuning.fit(X_train, y_train)

        y_pred = model_tuning.predict(X_val)

        f1_folds.append(
            f1_score(
                y_val,
                y_pred,
                average="macro",
                zero_division=0
            )
        )

        hamming_folds.append(
            hamming_loss(y_val, y_pred)
        )

    hasil_tuning.append([
        param["n_estimators"],
        param["max_depth"],
        param["min_samples_leaf"],
        np.mean(f1_folds),
        np.mean(hamming_folds)
    ])
```

```python
hasil_tuning_df = pd.DataFrame(
    hasil_tuning,
    columns=[
        "n_estimators",
        "max_depth",
        "min_samples_leaf",
        "F1_Macro",
        "Hamming_Loss"
    ]
)

hasil_tuning_df.sort_values(
    by="F1_Macro",
    ascending=False
).round(3)
```
